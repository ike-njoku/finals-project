Self Assessment Tool for Rehabilitation and TUG analysis
---
---
In this notebook, we explore the use of wearable IMU devices to create a self assessment system for rehabilitation and TUG analysis.  The devices are worn around the waist, around the knee and around the thigh.  
Each device records accelerometer and gyroscope measurements, as well as the arduino onboard clock timestamp (when the data was recorded). Each device has slight variations in the arduino milliseconds recorded.

#### The different Approaches 
---

To achieve our aim, we will experiment with different variations of our dataset, and different experimental approaches. Our resulting model will be a combination of the best performing dataset and approach combination. 

Because the timestamps from each IMU device is different, we will take different approaches to consolidate all the recorded measurements into a usable dataset

In the first approach, we will merge the data from each node into a dataset, using the average of the `pc_timestamp` field for each row. 

In the second approach, we will use each IMU device independently, flatten them and use them in building the models

```python
def print_project_tree(directory: Path, prefix: str = "", ignore_hidden: bool = False):
    """Recursively prints a visual tree structure of a directory."""
    entries = sorted(list(directory.iterdir()), key=lambda e: (e.is_file(), e.name.lower()))
    
    if ignore_hidden:
        entries = [e for e in entries if not e.name.startswith('.')]

    count = len(entries)
    for index, entry in enumerate(entries):
        is_last = (index == count - 1)
        connector = "└── " if is_last else "├── "
        
        print(f"{prefix}{connector}{entry.name}{'/' if entry.is_dir() else ''}")
        
        if entry.is_dir():
            extension = "    " if is_last else "│   "
            print_project_tree(entry, prefix=prefix + extension, ignore_hidden=ignore_hidden)

if __name__ == "__main__":
    # Point this to your project folder path (e.g., Path("."))
    data_folder = Path("./data") 
    print(f"{data_folder.resolve().name}/")
    print_project_tree(data_folder, ignore_hidden=True)
```

#### The DataSet
---
Each IMU device measures and records the following fields: 
| Field Name | Description |
| :--- | :--- |
| `pc_timestamp` | The absolute wall-clock timestamp recorded by the receiving computer/server when the data packet was received. |
| `arduiono_ms` | The relative uptime timestamp in milliseconds (`millis()`) recorded by the Arduino microcontroller at the exact time of sample capture. |
| `node_position` | The physical location or placement of the sensor node on the subject's body (e.g., wrist, waist, thigh). |
| `activity` | The ground-truth physical activity label being performed during data collection (e.g., sitting, standing, walking). |
| `ax` | Acceleration along the spatial $X$-axis measured by the accelerometer. |
| `ay` | Acceleration along the spatial $Y$-axis measured by the accelerometer. |
| `az` | Acceleration along the spatial $Z$-axis measured by the accelerometer. |
| `gx` | Angular velocity (rotational speed) around the spatial $X$-axis measured by the gyroscope. |
| `gy` | Angular velocity (rotational speed) around the spatial $Y$-axis measured by the gyroscope. |
| `gz` | Angular velocity (rotational speed) around the spatial $Z$-axis measured by the gyroscope. |
| `rssi` | Received Signal Strength Indicator (typically in dBm), measuring the wireless signal quality between the IMU node and the receiver. |


Building the datasets: There will be two datasets, in `Dataset A`, all the columns in the different datasets will be merged, using the `pc_timestamp`
For each user, the three different sensors will be merged into one dataset for each activity. 

so, for `user_1` where activity is sit, the knee, lumbar and thigh datasets will be merged into one.

```python
Dataset_A = pd.DataFrame()  # Initialize an empty DataFrame to hold the merged dataset

# load Dataset_A.csv if it already exists
if Path("Dataset_A.csv").exists():
    Dataset_A = pd.read_csv("Dataset_A.csv")
    print(f"Loaded existing Dataset_A.csv with shape: {Dataset_A.shape}")   

# Handle path whether running from project root or inside the 'notebooks/' directory
data_folder = Path("../data") if Path("../data").exists() else Path("data")

def load_and_prep_sensor(file_path: Path, node_label: str, join_key: str = "pc_timestamp") -> pd.DataFrame:
    " ""Reads CSV, parses timestamps, sorts chronologically, and prefixes feature columns."""
    df = pd.read_csv(file_path)
    
    # Convert timestamp column to datetime
    df[join_key] = pd.to_datetime(df[join_key])
    
    # Sort by timestamp required for pd.merge_asof
    df = df.sort_values(by=join_key).reset_index(drop=True)
    
    # Prefix feature columns with node label (e.g., knee_ax) to avoid column collisions
    rename_dict = {col: f"{node_label}_{col}" for col in df.columns if col != join_key}
    return df.rename(columns=rename_dict)

all_session_dfs = []

# Aggregate metrics trackers
metrics = {
    "knee_raw": 0,
    "lumbar_raw": 0,
    "thigh_raw": 0,
    "lumbar_mismatches": 0,
    "thigh_mismatches": 0,
    "final_rows": 0
}

# Create a formatted table header to keep session metrics
print(f"{'Session':<22} | {'Knee':<6} | {'Lumbar':<6} | {'Thigh':<6} | {'L_Mismatch':<10} | {'T_Mismatch':<10} | {'Final':<6}")
print("-" * 80)

# Iterate through user folders (user_1, user_2, etc.)
for user_dir in sorted(data_folder.glob("user_*")):
    if not user_dir.is_dir():
        continue

    # Iterate through activity folders (sit, stand, tug, walk)
    for activity_dir in sorted(user_dir.iterdir()):
        if not activity_dir.is_dir():
            continue

        knee_file = activity_dir / "session-knee.csv"
        lumbar_file = activity_dir / "session-lumbar.csv"
        thigh_file = activity_dir / "session-thigh.csv"

        # Ensure all three node files exist
        if not (knee_file.exists() and lumbar_file.exists() and thigh_file.exists()):
            print(f"Skipping {user_dir.name}/{activity_dir.name}: Missing file(s).")
            continue

        # Load, convert timestamps, sort, and prefix columns
        df_knee = load_and_prep_sensor(knee_file, "knee")
        df_lumbar = load_and_prep_sensor(lumbar_file, "lumbar")
        df_thigh = load_and_prep_sensor(thigh_file, "thigh")

        raw_knee_count = len(df_knee)
        raw_lumbar_count = len(df_lumbar)
        raw_thigh_count = len(df_thigh)

        # Step 1: Align Lumbar to Knee (anchor) within a 50ms tolerance window
        merged = pd.merge_asof(
            df_knee,
            df_lumbar,
            on="pc_timestamp",
            direction="nearest",
            tolerance=pd.Timedelta("50ms")
        )

        #  Align Thigh to the combined dataset within a 50ms tolerance window
        merged = pd.merge_asof(
            merged,
            df_thigh,
            on="pc_timestamp",
            direction="nearest",
            tolerance=pd.Timedelta("50ms")
        )

        # Identify sample sensor columns to detect NaNs resulting from >50ms mismatches
        lumbar_col = [c for c in merged.columns if c.startswith("lumbar_")][0]
        thigh_col = [c for c in merged.columns if c.startswith("thigh_")][0]

        lumbar_mismatch = merged[lumbar_col].isna().sum()
        thigh_mismatch = merged[thigh_col].isna().sum()

        # Drop rows with unmatched missing values to ensure a complete multi-node dataset
        session_df = merged.dropna().reset_index(drop=True)
        final_count = len(session_df)

        # Add context metadata
        session_df["user_id"] = user_dir.name
        session_df["activity"] = activity_dir.name
        all_session_dfs.append(session_df)

        # Update totals
        metrics["knee_raw"] += raw_knee_count
        metrics["lumbar_raw"] += raw_lumbar_count
        metrics["thigh_raw"] += raw_thigh_count
        metrics["lumbar_mismatches"] += lumbar_mismatch
        metrics["thigh_mismatches"] += thigh_mismatch
        metrics["final_rows"] += final_count

        session_name = f"{user_dir.name}/{activity_dir.name}"
        print(f"{session_name:<22} | {raw_knee_count:<6} | {raw_lumbar_count:<6} | {raw_thigh_count:<6} | {lumbar_mismatch:<10} | {thigh_mismatch:<10} | {final_count:<6}")

# Combine all merged sessions into final dataset
if all_session_dfs:
    Dataset_A = pd.concat(all_session_dfs, ignore_index=True)
    Dataset_A.to_csv("Dataset_A.csv", index=False)

    print("=" * 80)
    print("DATASET DROPPED ROWS & MISMATCH SUMMARY")
    print("=" * 80)
    print(f"Total Raw Knee Rows Loaded   : {metrics['knee_raw']}")
    print(f"Total Raw Lumbar Rows Loaded : {metrics['lumbar_raw']}")
    print(f"Total Raw Thigh Rows Loaded  : {metrics['thigh_raw']}")
    print("-" * 50)
    print(f"Lumbar Mismatches (>50ms)    : {metrics['lumbar_mismatches']} rows dropped")
    print(f"Thigh Mismatches (>50ms)     : {metrics['thigh_mismatches']} rows dropped")
    print("-" * 50)
    print(f"Total Rows Dropped from Knee   : {metrics['knee_raw'] - metrics['final_rows']} ({((metrics['knee_raw'] - metrics['final_rows'])/metrics['knee_raw'])*100:.2f}%)")
    print(f"Total Rows Dropped from Lumbar : {metrics['lumbar_raw'] - metrics['final_rows']} ({((metrics['lumbar_raw'] - metrics['final_rows'])/metrics['lumbar_raw'])*100:.2f}%)")
    print(f"Total Rows Dropped from Thigh  : {metrics['thigh_raw'] - metrics['final_rows']} ({((metrics['thigh_raw'] - metrics['final_rows'])/metrics['thigh_raw'])*100:.2f}%)")
    print("-" * 50)
    print(f"Final Synchronized Rows      : {metrics['final_rows']}")
    print(f"Final Output Shape           : {Dataset_A.shape}")
    print("Saved merged dataset to 'merged_dataset.csv'")
else:
    print("No matching session CSV files were found. Verify your current working directory.")
```

Our dataset is comprised of thousands of snapshots of readings from accelerometers x, y and z, together with the timestamp at which that snapshot was recorded.
While these individual snapshots are important, they do not tell the full story of what activity is being perormed. This is because realisitically, activities are a set of fluid motions (or not) that are interlinked, performed over time. 

So, to see if someone is currently running, we need to see for the past x seconds, what movements the person was performing. The x-seconds, is called a window. And or the sake of consistency, we will keep the window size constant. 

The dataset was collected at roughly 20 samples/second (20Hz) (check the frequency distributions plotted or the x, y, and z columns above) [.]. We will use 5 seconds as our window, giving us a total of 100 samples or snapshots with which to classifify activities.  

Window 1:
[
 [x1,y1,z1],
 [x2,y2,z2],
 ...
 [x100,y100,z100]
] ==> Walking


Window 2:
[
 [x1,y1,z1],
 [x2,y2,z2],
 ...
 [x100,y100,z100]
] ==> Jogging

Window 3:
[
 [x1,y1,z1],
 [x2,y2,z2],
 ...
 [x100,y100,z100]
] ==> Walking

It is also important to introduce an overlap between the windows, so that the model does not miss out on interesting information stored in at the edges of the window. We will use an overlap that is 50% of the window samples. 
While this helps paint a better picture of the activity being performed, It also translates to longer training times and more memory consumption.

##### Did every user perform every activity?
---
User-level analysis is particularly important in human activity recognition because individuals exhibit unique movement patterns. To avoid data leakage and ensure generalizability, subject-independent evaluation methods such as leave-one-subject-out validation should be considered rather than random train-test splits.

```python
#check if each user performed all activities
check_list = pd.crosstab(
    Dataset_A["user_id"],
    Dataset_A["activity"]
)
# Convert sample counts into binary flags: 1 if they did it at all, 0 if completely missing
completion_matrix = (check_list > 0).astype(int)

plt.figure(figsize=(24, 8))
sns.heatmap(
    completion_matrix, 
    cmap="Blues", 
    cbar=False, 
    linewidths=0.5, 
    linecolor='lightgray'
)

plt.title("Activity Checksheet: Did the User Perform This Task?", fontsize=14, fontweight='bold', pad=15)
plt.ylabel("User ID", fontsize=12)
plt.xlabel("Activity Status", fontsize=12)
plt.tight_layout()
plt.show()
```

##### Comparison of accelerator values for different activities performed by one user (who completed all activities)
---

```python
def get_user_activity_data(df, user_id, activity, num_samples=200):
    """Filters data for a specific user and activity, and returns a small sample."""
    sample = df[(df['user_id'] == f"user_{user_id}") & (df['activity'] == activity)].head(num_samples)
    return sample    

def plot_user_activity_sample(sample_df, title, ax, sensor, columns=None):
    '''Plots all 9 IMU sensor channels for a specific user and activity.'''
    if columns is None:
        
        if sensor == "accelerometer": 

            columns = [
                'knee_ax', 'knee_ay', 'knee_az',
                'thigh_ax', 'thigh_ay', 'thigh_az',
                'lumbar_ax', 'lumbar_ay', 'lumbar_az'
            ]
        else: 
            columns = [
                        'knee_gx', 'knee_gy', 'knee_gz',
                        'thigh_gx', 'thigh_gy', 'thigh_gz',
                        'lumbar_gx', 'lumbar_gy', 'lumbar_gz'
                    ]

    # Map axis to color and body node to line style

    color_map = {'ax': 'red', 'ay': 'blue', 'az': 'green','gx': 'red', 'gy': 'blue', 'gz': 'green'}
    style_map = {'knee': '-', 'thigh': '--', 'lumbar': ':'}

    for col in columns:
        if col in sample_df.columns:
            node, axis = col.split('_')  # Extracts 'knee' and 'ax'
            
            ax.plot(
                sample_df.index,
                sample_df[col],
                label=f"{node.capitalize()} {axis.upper()}",
                color=color_map.get(axis, 'black'),
                linestyle=style_map.get(node, '-'),
                alpha=0.8
            )

    ax.set_title(title, fontsize=14, fontweight='bold')
    ax.set_ylabel('Acceleration (g)', fontsize=12)
    
    # Place legend outside the plot so it doesn't cover the data
    ax.legend(loc='upper left', bbox_to_anchor=(1.02, 1), borderaxespad=0)

def get_sample_user_data(user_id):

    sample_user_walk = get_user_activity_data(Dataset_A, user_id=user_id, activity='walk', num_samples=200) 
    sample_user_stand = get_user_activity_data(Dataset_A, user_id=user_id, activity='stand', num_samples=200)
    sample_user_sit = get_user_activity_data(Dataset_A, user_id=user_id, activity='sit', num_samples=200)
    sample_user_tug = get_user_activity_data(Dataset_A, user_id=user_id, activity='tug', num_samples=200)
    
    return (sample_user_walk, sample_user_stand, sample_user_sit, sample_user_tug)


sample_user_walk, sample_user_stand, sample_user_sit, sample_user_tug = get_sample_user_data(SAMPLE_USER)

fig, (ax1, ax2, ax3, ax4) = plt.subplots(nrows=4, ncols=1, figsize=(24, 24))

plot_user_activity_sample(sample_user_stand, "Sample User Stand", ax1, sensor="accelerometer")
plot_user_activity_sample(sample_user_sit, "Sample User Sit", ax2, sensor="accelerometer")
plot_user_activity_sample(sample_user_walk, "Sample User Walk", ax3, sensor="accelerometer");
plot_user_activity_sample(sample_user_tug, "Sample User Timed Up and Go", ax4, sensor="accelerometer")
```

As shown in the plot above, the patterns recorded by the accelerometers differ with respect to activity. While walking, stairs-ascent and decent and jogging produce regular heavy spikes in the readings of the accelerometers, sitting and standing generally experience less violent spikes, and they happen far less frequently.

#### Data Preprocessing
---

Human activity usually occurs at around 3.0Hz to 4.0Hz. Because of this, it is safe to assume that accelerometers that fall outside of this may be noise generated by the sensors and the effects of movements. 

To make sure that we are tracking only human activity, we need to filter out the noise. Below, we use a low pass filter, keeping our Nyquist frequency at 10Hz (half of our accelerometer sampling rates)

##### Applying a low-pass filter

```python
from scipy.signal import butter, filtfilt
print_horizontal_line()

print(f'{len(Dataset_A)} total samples before filtering')
       
print_horizontal_line()
# Define the filter properties
# The WISDM dataset is typically sampled at 20Hz. We set a cut-off at 4.0Hz 
# to capture human movement while filtering out high-frequency noise/vibrations.
def butter_lowpass_filter(data, cutoff=4.0, fs=20.0, order=4):
    nyq = 0.5 * fs  # Nyquist Frequency (absolute limit of the sensors, where readings can be captured without distortion)
    normal_cutoff = cutoff / nyq
    b, a = butter(order, normal_cutoff, btype='low', analog=False)
    # filtfilt applies the filter forward and backward to ensure zero-phase shift
    return filtfilt(b, a, data)

# Apply to your existing dataframe columns
# Make sure to run this *after* your dropna() / drop_duplicates() steps
for axis in columns_to_plot:
    Dataset_A[axis] = butter_lowpass_filter(Dataset_A[axis].values) # assign the filters

print(f'{len(Dataset_A)} total samples after filtering')
print_horizontal_line()
```

##### Feature Engineering
___
While the x,y and z components are very important in activity classification, there are other pieces of information that are often needed. These bits of infomration need to be derived from the accelerometer values. 

1. Signal Magnitude Vector: The total acceleration experienced by phone, regardless of direction.
   [Magnitude = √(x² + y² + z²)]
2. Roll: How much the phone is rotated sideways [Roll = atan2(y, z) × (180/π)]
3. Pitch: Pitch measures forward and backward tilt. Pitch = atan2(
    -x,
    √(y² + z²)
)
4. Jerk: How quickly acceleration changes. Jerk_x = (x_t - x_(t-1)) / Δt
5. Magnitude of Jerk: Overall rate of change of acceleration
6. Mean acceleration: The average acceleration over a time window
   

##### Accelerometer Engineered features
---

```python
def add_accelerometer_features(
    df,
    nodes=("knee", "thigh", "lumbar"),
    sampling_rate_hz=SAMPLING_RATE_HZ,
    rolling_window_seconds=5.0
):
    """
    Creates accelerometer-derived features without modifying the original DataFrame.

    Features are calculated independently within contiguous activity sessions
    for each participant.

    Parameters
    ----------
    df : pd.DataFrame
        Input dataset.

    nodes : tuple
        IMU node prefixes.

    sampling_rate_hz : int
        Sensor sampling frequency in Hz.

    rolling_window_seconds : float
        Duration of rolling window in seconds.

    Returns
    -------
    pd.DataFrame
        Copy of the dataset with accelerometer-derived features.
    """

    data = df.copy()

    dt = 1.0 / sampling_rate_hz

    rolling_window_samples = int(
        rolling_window_seconds * sampling_rate_hz
    )

    # -----------------------------------------------------------------------
    # SORT DATA CHRONOLOGICALLY
    # -----------------------------------------------------------------------

    time_candidates = [
        "pc_timestamp",
    ]

    time_col = next(
        (col for col in time_candidates if col in data.columns),
        None
    )

    if time_col is not None:
        data = data.sort_values(
            ["user_id", time_col]
        ).reset_index(drop=True)

    # -----------------------------------------------------------------------
    # CREATE CONTIGUOUS SESSION IDS
    # -----------------------------------------------------------------------

    data["activity_clean"] = (
        data["activity"]
        .astype(str)
        .str.strip()
        .str.lower()
    )

    data["session_id"] = (
        (data["user_id"] != data["user_id"].shift())
        |
        (data["activity_clean"] != data["activity_clean"].shift())
    ).cumsum()

    # -----------------------------------------------------------------------
    # FEATURE ENGINEERING
    # -----------------------------------------------------------------------

    for node in nodes:

        ax = f"{node}_ax"
        ay = f"{node}_ay"
        az = f"{node}_az"

        required_cols = [ax, ay, az]

        if not all(col in data.columns for col in required_cols):

            print(
                f"Skipping accelerometer features for '{node}'. "
                f"Missing one or more columns: {required_cols}"
            )

            continue

        # -------------------------------------------------------------------
        # 1. ACCELERATION MAGNITUDE
        # -------------------------------------------------------------------

        data[f"{node}_accel_mag"] = np.sqrt(
            data[ax] ** 2
            +
            data[ay] ** 2
            +
            data[az] ** 2
        )

        # -------------------------------------------------------------------
        # 2. ROLL
        # -------------------------------------------------------------------

        data[f"{node}_roll"] = np.degrees(
            np.arctan2(
                data[ay],
                data[az]
            )
        )

        # -------------------------------------------------------------------
        # 3. PITCH
        # -------------------------------------------------------------------

        data[f"{node}_pitch"] = np.degrees(
            np.arctan2(
                -data[ax],
                np.sqrt(
                    data[ay] ** 2
                    +
                    data[az] ** 2
                )
            )
        )

        # -------------------------------------------------------------------
        # 4. JERK
        #
        # IMPORTANT:
        # Calculated inside each contiguous session so the derivative does not
        # cross from one activity/user into another.
        # -------------------------------------------------------------------

        data[f"{node}_jerk_x"] = (
            data.groupby("session_id")[ax]
            .diff()
            / dt
        )

        data[f"{node}_jerk_y"] = (
            data.groupby("session_id")[ay]
            .diff()
            / dt
        )

        data[f"{node}_jerk_z"] = (
            data.groupby("session_id")[az]
            .diff()
            / dt
        )

        data[f"{node}_jerk_mag"] = np.sqrt(
            data[f"{node}_jerk_x"] ** 2
            +
            data[f"{node}_jerk_y"] ** 2
            +
            data[f"{node}_jerk_z"] ** 2
        )

        # -------------------------------------------------------------------
        # 5. ENERGY
        # -------------------------------------------------------------------

        data[f"{node}_energy_x"] = data[ax] ** 2
        data[f"{node}_energy_y"] = data[ay] ** 2
        data[f"{node}_energy_z"] = data[az] ** 2

        data[f"{node}_accel_energy"] = (
            data[ax] ** 2
            +
            data[ay] ** 2
            +
            data[az] ** 2
        )

        # -------------------------------------------------------------------
        # 6. ROLLING MEAN ACCELERATION
        # -------------------------------------------------------------------

        data[f"{node}_window_mean_acc"] = (
            data.groupby("session_id")[f"{node}_accel_mag"]
            .transform(
                lambda x: x.rolling(
                    window=rolling_window_samples,
                    min_periods=1
                ).mean()
            )
        )

    # -----------------------------------------------------------------------
    # FILL NaN VALUES CREATED BY DIFFERENTIATION
    # -----------------------------------------------------------------------

    jerk_cols = [
        col
        for col in data.columns
        if "jerk" in col
    ]

    data[jerk_cols] = data[jerk_cols].fillna(0)

    return data
```

##### Gyroscope Engineered features
---

```python
def add_gyroscope_features(
    df,
    nodes=("knee", "thigh", "lumbar"),
    sampling_rate_hz=SAMPLING_RATE_HZ,
    rolling_window_seconds=2.5
):
    """
    Adds gyroscope-derived features.

    Features are calculated inside contiguous activity sessions to prevent
    derivatives and rolling windows from crossing user/activity boundaries.
    """

    data = df.copy()

    dt = 1.0 / sampling_rate_hz

    rolling_window_samples = int(
        rolling_window_seconds * sampling_rate_hz
    )

    # -----------------------------------------------------------------------
    # ENSURE SESSION ID EXISTS
    # -----------------------------------------------------------------------

    if "session_id" not in data.columns:

        data["activity_clean"] = (
            data["activity"]
            .astype(str)
            .str.strip()
            .str.lower()
        )

        data["session_id"] = (
            (data["user_id"] != data["user_id"].shift())
            |
            (
                data["activity_clean"]
                !=
                data["activity_clean"].shift()
            )
        ).cumsum()

    # -----------------------------------------------------------------------
    # GYROSCOPE FEATURES
    # -----------------------------------------------------------------------

    for node in nodes:

        gx = f"{node}_gx"
        gy = f"{node}_gy"
        gz = f"{node}_gz"

        ax = f"{node}_ax"
        ay = f"{node}_ay"
        az = f"{node}_az"

        gyro_cols = [gx, gy, gz]

        if not all(col in data.columns for col in gyro_cols):

            print(
                f"Skipping gyroscope features for '{node}'."
            )

            continue

        # -------------------------------------------------------------------
        # 1. ANGULAR VELOCITY MAGNITUDE
        # -------------------------------------------------------------------

        data[f"{node}_gyro_mag"] = np.sqrt(
            data[gx] ** 2
            +
            data[gy] ** 2
            +
            data[gz] ** 2
        )

        # -------------------------------------------------------------------
        # 2. ANGULAR ACCELERATION
        # -------------------------------------------------------------------

        data[f"{node}_ang_acc_x"] = (
            data.groupby("session_id")[gx]
            .diff()
            / dt
        )

        data[f"{node}_ang_acc_y"] = (
            data.groupby("session_id")[gy]
            .diff()
            / dt
        )

        data[f"{node}_ang_acc_z"] = (
            data.groupby("session_id")[gz]
            .diff()
            / dt
        )

        # -------------------------------------------------------------------
        # 3. GYROSCOPE ENERGY
        # -------------------------------------------------------------------

        data[f"{node}_gyro_energy"] = (
            data[gx] ** 2
            +
            data[gy] ** 2
            +
            data[gz] ** 2
        )

        # -------------------------------------------------------------------
        # 4. ROLLING STANDARD DEVIATION
        # -------------------------------------------------------------------

        data[f"{node}_gyro_mag_std"] = (
            data.groupby("session_id")[f"{node}_gyro_mag"]
            .transform(
                lambda x: x.rolling(
                    window=rolling_window_samples,
                    min_periods=1
                ).std()
            )
            .fillna(0)
        )

        # -------------------------------------------------------------------
        # 5. APPROXIMATE ANGULAR DISPLACEMENT
        # -------------------------------------------------------------------

        data[f"{node}_angle_disp_z"] = (
            data.groupby("session_id")[gz]
            .transform(
                lambda x: (
                    x * dt
                ).rolling(
                    window=rolling_window_samples,
                    min_periods=1
                ).sum()
            )
        )

        # -------------------------------------------------------------------
        # 6. MOTION ALIGNMENT
        # -------------------------------------------------------------------

        accel_cols = [ax, ay, az]

        if all(col in data.columns for col in accel_cols):

            accel_mag = np.sqrt(
                data[ax] ** 2
                +
                data[ay] ** 2
                +
                data[az] ** 2
            )

            gyro_mag = data[f"{node}_gyro_mag"]

            dot_product = (
                data[ax] * data[gx]
                +
                data[ay] * data[gy]
                +
                data[az] * data[gz]
            )

            data[f"{node}_motion_alignment"] = (
                dot_product
                /
                (accel_mag * gyro_mag + 1e-8)
            )

    # -----------------------------------------------------------------------
    # CLEAN DERIVATIVE NaNs
    # -----------------------------------------------------------------------

    angular_acc_cols = [
        col
        for col in data.columns
        if "ang_acc" in col
    ]

    data[angular_acc_cols] = (
        data[angular_acc_cols]
        .fillna(0)
    )

    return data
```

```python
# =============================================================================
# CELL 68: DATA PREPARATION FOR SIT-TO-STAND AND WALKING ANALYSIS
# =============================================================================
#
# PURPOSE
# -------
#
# This cell prepares Dataset_A for the analyses performed in Cells 69 and 70.
#
# Specifically, it:
#
#   1. Creates a safe working copy of Dataset_A.
#   2. Checks that the required raw IMU and metadata columns exist.
#   3. Calculates the three-dimensional knee acceleration magnitude.
#   4. Normalises activity labels.
#   5. Sorts the data chronologically within each participant.
#
#
# IMPORTANT
# ---------
#
# knee_accel_mag is NOT assumed to already exist in Dataset_A.
#
# It is calculated directly from:
#
#       knee_ax
#       knee_ay
#       knee_az
#
#
# Formula:
#
#                         _________________________________
#                        /          2          2          2
#       |a| =           √  knee_ax + knee_ay + knee_az
#
#
# The resulting df DataFrame will be used by Cell 69.
# =============================================================================


# =============================================================================
# 1. IMPORT REQUIRED LIBRARIES
# =============================================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt


# =============================================================================
# 2. CREATE A WORKING COPY OF THE MASTER DATASET
# =============================================================================
#
# We use a copy so that the original Dataset_A remains unchanged.
# =============================================================================

df = Dataset_A.copy()


# =============================================================================
# 3. DEFINE THE REQUIRED RAW COLUMNS
# =============================================================================
#
# These are the columns required for the analyses.
#
# We do NOT include:
#
#       knee_accel_mag
#
# because it is created below.
# =============================================================================

required_columns = [
    "user_id",
    "activity",
    "knee_ax",
    "knee_ay",
    "knee_az"
]


# =============================================================================
# 4. CHECK FOR MISSING COLUMNS
# =============================================================================

missing_columns = [
    column
    for column in required_columns
    if column not in df.columns
]


if missing_columns:

    raise ValueError(
        "The following required columns are missing: "
        f"{missing_columns}"
    )


# =============================================================================
# 5. CALCULATE 3D KNEE ACCELERATION MAGNITUDE
# =============================================================================
#
# The three accelerometer axes are combined into one acceleration magnitude.
#
# This produces a single signal representing the overall acceleration
# experienced by the knee-mounted IMU.
# =============================================================================

df["knee_accel_mag"] = np.sqrt(
    df["knee_ax"] ** 2
    +
    df["knee_ay"] ** 2
    +
    df["knee_az"] ** 2
)

# =============================================================================
# 7. SORT THE DATA CHRONOLOGICALLY
# =============================================================================
#
# If pc_timestamp is available, sort each participant's data chronologically.
#
# This is important because:
#
#   - sit -> stand transitions depend on sequence order;
#   - step detection requires walking samples to remain in chronological order.
# =============================================================================

if "pc_timestamp" in df.columns:

    df = (
        df
        .sort_values(
            by=[
                "user_id",
                "pc_timestamp"
            ]
        )
        .reset_index(drop=True)
    )

else:

    # -------------------------------------------------------------------------
    # If no timestamp exists, preserve the original ordering within each user.
    # -------------------------------------------------------------------------

    df = (
        df
        .sort_values(
            by="user_id"
        )
        .reset_index(drop=True)
    )


# =============================================================================
# 8. DISPLAY A PREPARATION SUMMARY
# =============================================================================

print("=" * 80)
print("DATA PREPARATION COMPLETE")
print("=" * 80)

print(
    f"\nNumber of samples: {len(df):,}"
)

print(
    f"Number of individuals: {df['user_id'].nunique()}"
)

print(
    "\nAvailable activity labels:"
)

print(
    sorted(
        df["activity"]
        .dropna()
        .unique()
    )
)

print(
    "\nColumns created in this cell:"
)

print(
    [
        "knee_accel_mag",
        "activity"
    ]
)


# =============================================================================
# END OF CELL 68
# =============================================================================
```

```python
# =============================================================================
# CELL 69: SIT-TO-STAND TIME AND ESTIMATED WALKING DISTANCE ANALYSIS
# =============================================================================
#
# THIS CELL USES THE DATA PREPARED IN CELL 68.
#
#
# ANALYSIS 1: SIT-TO-STAND TRANSITION TIME
# -----------------------------------------
#
# The code:
#
#   1. Finds contiguous activity sessions.
#   2. Detects transitions from "sit" to "stand".
#   3. Examines knee acceleration variability around each transition.
#   4. Identifies the active movement period.
#   5. Calculates the duration of the detected movement.
#   6. Calculates average sit-to-stand time for each individual.
#   7. Calculates the average sit-to-stand time across individuals.
#
#
# ANALYSIS 2: WALKING DISTANCE ESTIMATION
# ---------------------------------------
#
# The dataset does not contain directly measured walking distance.
#
# Therefore, distance is estimated using:
#
#       Estimated Distance = Number of Detected Steps × Step Length
#
#
# The process is:
#
#   1. Extract walking samples.
#   2. Separate independent walking sessions.
#   3. Detect peaks corresponding to potential steps.
#   4. Calculate step count.
#   5. Estimate distance using a configurable average step length.
#   6. Calculate total walking distance per individual.
#   7. Calculate average walking distance across individuals.
#
#
# IMPORTANT METHODOLOGICAL NOTE
# -----------------------------
#
# The resulting walking distance is an ESTIMATE.
#
# A constant step length is used because participant-specific anthropometric
# measurements or ground-truth step lengths are not currently available in
# the dataset.
#
# =============================================================================


# =============================================================================
# 1. ANALYSIS CONFIGURATION
# =============================================================================

# -------------------------------------------------------------------------
# Your stated IMU sampling frequency.
# -------------------------------------------------------------------------

SAMPLING_RATE_HZ = 20


# -------------------------------------------------------------------------
# SIT-TO-STAND ANALYSIS PARAMETERS
# -------------------------------------------------------------------------

# Total window examined around each labelled sit -> stand boundary.

SIT_STAND_WINDOW_SECONDS = 6.0


# Duration of the rolling window used to calculate local acceleration
# variability.

VARIANCE_WINDOW_SECONDS = 0.5


# Adaptive movement threshold:
#
#     threshold = median + (multiplier × MAD)
#
# MAD = Median Absolute Deviation.

SIT_STAND_THRESHOLD_MULTIPLIER = 3.0


# Physiologically plausible limits for detected sit-to-stand movements.

MIN_SIT_STAND_DURATION_SECONDS = 0.5
MAX_SIT_STAND_DURATION_SECONDS = 6.0


# -------------------------------------------------------------------------
# WALKING / STEP DETECTION PARAMETERS
# -------------------------------------------------------------------------

# Minimum time expected between consecutive steps.
#
# 0.30 seconds corresponds to a maximum theoretical cadence of approximately
# 200 steps per minute.

MIN_STEP_INTERVAL_SECONDS = 0.30


# Minimum peak prominence is determined adaptively from the walking signal.

STEP_PROMINENCE_MULTIPLIER = 0.50


# -------------------------------------------------------------------------
# ESTIMATED STEP LENGTH
# -------------------------------------------------------------------------
#
# This is a configurable population-level assumption.
#
# It is deliberately named ESTIMATED_STEP_LENGTH_METERS so that the code
# does not imply that this value was directly measured.
#
# You may later replace this with:
#
#   - participant-specific step lengths;
#   - height-based estimates;
#   - stride length estimated from validated gait models;
#   - experimentally measured ground truth.
# -------------------------------------------------------------------------

ESTIMATED_STEP_LENGTH_METERS = 0.70


# =============================================================================
# 2. CREATE CONTIGUOUS ACTIVITY SESSIONS
# =============================================================================
#
# A new activity session starts whenever:
#
#   1. The participant changes, OR
#   2. The activity changes.
#
#
# Example:
#
# user_1:
#
#     sit
#     sit
#     sit
#     stand
#     stand
#     walk
#     walk
#
# becomes:
#
#     session 1 -> sit
#     session 2 -> stand
#     session 3 -> walk
#
# =============================================================================

df["session_boundary"] = (
    (df["user_id"] != df["user_id"].shift())
    |
    (
        df["activity"]
        !=
        df["activity"].shift()
    )
)


df["analysis_session_id"] = (
    df["session_boundary"]
    .cumsum()
)


# =============================================================================
# 3. CREATE A SUMMARY OF EACH ACTIVITY SESSION
# =============================================================================

session_summary = (
    df
    .groupby(
        [
            "user_id",
            "analysis_session_id"
        ],
        sort=False
    )
    .agg(
        activity=(
            "activity",
            "first"
        ),

        start_index=(
            "activity",
            lambda x: x.index.min()
        ),

        end_index=(
            "activity",
            lambda x: x.index.max()
        ),

        number_of_samples=(
            "activity",
            "size"
        )
    )
    .reset_index()
)


# =============================================================================
# 4. IDENTIFY SIT -> STAND TRANSITIONS
# =============================================================================

session_summary["previous_activity"] = (
    session_summary
    .groupby("user_id")["activity"]
    .shift(1)
)


sit_to_stand_boundaries = (
    session_summary[
        (
            session_summary["previous_activity"] == "sit"
        )
        &
        (
            session_summary["activity"] == "stand"
        )
    ]
    .copy()
)


print("=" * 80)
print("SIT-TO-STAND TRANSITION DETECTION")
print("=" * 80)

print(
    f"\nNumber of labelled sit -> stand transitions found: "
    f"{len(sit_to_stand_boundaries)}"
)


# =============================================================================
# 5. CONVERT TIME WINDOWS TO SAMPLES
# =============================================================================

half_sit_stand_window_samples = int(
    (SIT_STAND_WINDOW_SECONDS / 2)
    *
    SAMPLING_RATE_HZ
)


variance_window_samples = max(
    3,
    int(
        VARIANCE_WINDOW_SECONDS
        *
        SAMPLING_RATE_HZ
    )
)


# =============================================================================
# 6. ANALYSE EACH SIT -> STAND TRANSITION
# =============================================================================

sit_to_stand_records = []


for _, transition in sit_to_stand_boundaries.iterrows():

    # ---------------------------------------------------------------------
    # Identify the current participant.
    # ---------------------------------------------------------------------

    current_user = transition["user_id"]


    # ---------------------------------------------------------------------
    # The beginning of the stand session represents the labelled boundary.
    # ---------------------------------------------------------------------

    transition_index = int(
        transition["start_index"]
    )


    # ---------------------------------------------------------------------
    # Define a local analysis window around the transition.
    # ---------------------------------------------------------------------

    start_window = max(
        0,
        transition_index - half_sit_stand_window_samples
    )


    end_window = min(
        len(df) - 1,
        transition_index + half_sit_stand_window_samples
    )


    # ---------------------------------------------------------------------
    # Extract the local window.
    # ---------------------------------------------------------------------

    local_window = (
        df
        .loc[
            start_window:end_window
        ]
        .copy()
    )


    # ---------------------------------------------------------------------
    # Ensure that the window contains only the current participant.
    # ---------------------------------------------------------------------

    local_window = (
        local_window[
            local_window["user_id"] == current_user
        ]
        .copy()
    )


    # ---------------------------------------------------------------------
    # Skip windows that are too short.
    # ---------------------------------------------------------------------

    if len(local_window) < variance_window_samples:
        continue


    # =====================================================================
    # CALCULATE LOCAL KNEE ACCELERATION VARIANCE
    # =====================================================================
    #
    # During relatively stationary sitting or standing, the signal should
    # generally exhibit lower variability.
    #
    # During the transition, movement causes increased acceleration changes.
    # =====================================================================

    local_window["motion_variance"] = (
        local_window["knee_accel_mag"]
        .rolling(
            window=variance_window_samples,
            center=True,
            min_periods=1
        )
        .var()
        .fillna(0)
    )


    # =====================================================================
    # CALCULATE AN ADAPTIVE MOVEMENT THRESHOLD
    # =====================================================================

    median_variance = (
        local_window["motion_variance"]
        .median()
    )


    mad_variance = np.median(
        np.abs(
            local_window["motion_variance"]
            -
            median_variance
        )
    )


    movement_threshold = (
        median_variance
        +
        (
            SIT_STAND_THRESHOLD_MULTIPLIER
            *
            mad_variance
        )
    )


    # Handle a completely flat MAD value.

    if mad_variance == 0:

        movement_threshold = (
            median_variance
            +
            1e-9
        )


    # =====================================================================
    # IDENTIFY ACTIVE MOVEMENT SAMPLES
    # =====================================================================

    local_window["is_active"] = (
        local_window["motion_variance"]
        >
        movement_threshold
    )


    # =====================================================================
    # IDENTIFY CONTIGUOUS ACTIVE MOVEMENT SEGMENTS
    # =====================================================================

    local_window["movement_group"] = (
        local_window["is_active"]
        .ne(
            local_window["is_active"].shift()
        )
        .cumsum()
    )


    candidate_segments = []


    for _, segment in local_window.groupby(
        "movement_group"
    ):

        # Ignore periods classified as inactive.

        if not segment["is_active"].iloc[0]:
            continue


        segment_start = int(
            segment.index.min()
        )


        segment_end = int(
            segment.index.max()
        )


        # Calculate how close this segment is to the labelled transition.

        if (
            segment_start
            <=
            transition_index
            <=
            segment_end
        ):

            distance_from_transition = 0

        else:

            distance_from_transition = min(
                abs(
                    segment_start
                    -
                    transition_index
                ),

                abs(
                    segment_end
                    -
                    transition_index
                )
            )


        candidate_segments.append(
            {
                "start_index": segment_start,
                "end_index": segment_end,
                "distance_from_transition": distance_from_transition
            }
        )


    # Skip if no active movement was identified.

    if len(candidate_segments) == 0:
        continue


    # Select the movement segment closest to the labelled transition.

    selected_segment = min(
        candidate_segments,
        key=lambda segment: segment[
            "distance_from_transition"
        ]
    )


    movement_start_index = (
        selected_segment["start_index"]
    )


    movement_end_index = (
        selected_segment["end_index"]
    )


    # =====================================================================
    # CALCULATE MOVEMENT DURATION
    # =====================================================================

    movement_samples = (
        movement_end_index
        -
        movement_start_index
        +
        1
    )


    sit_to_stand_seconds = (
        movement_samples
        /
        SAMPLING_RATE_HZ
    )


    # =====================================================================
    # APPLY PLAUSIBILITY FILTER
    # =====================================================================

    if (
        MIN_SIT_STAND_DURATION_SECONDS
        <=
        sit_to_stand_seconds
        <=
        MAX_SIT_STAND_DURATION_SECONDS
    ):

        sit_to_stand_records.append(
            {
                "user_id": current_user,
                "transition_index": transition_index,
                "movement_start_index": movement_start_index,
                "movement_end_index": movement_end_index,
                "movement_samples": movement_samples,
                "sit_to_stand_seconds": sit_to_stand_seconds,
                "movement_threshold": movement_threshold
            }
        )


# =============================================================================
# 7. CREATE SIT-TO-STAND RESULTS
# =============================================================================

sit_to_stand_results = pd.DataFrame(
    sit_to_stand_records
)


# =============================================================================
# 8. CALCULATE AVERAGE SIT-TO-STAND TIME PER INDIVIDUAL
# =============================================================================

if sit_to_stand_results.empty:

    average_sit_to_stand_per_user = pd.DataFrame(
        columns=[
            "user_id",
            "number_of_transitions",
            "average_sit_to_stand_seconds",
            "standard_deviation_seconds"
        ]
    )

    overall_average_sit_to_stand = np.nan


else:

    average_sit_to_stand_per_user = (
        sit_to_stand_results
        .groupby("user_id")
        .agg(
            number_of_transitions=(
                "sit_to_stand_seconds",
                "count"
            ),

            average_sit_to_stand_seconds=(
                "sit_to_stand_seconds",
                "mean"
            ),

            standard_deviation_seconds=(
                "sit_to_stand_seconds",
                "std"
            )
        )
        .reset_index()
    )


    # Calculate the overall average from individual averages.
    #
    # This gives every participant equal weight.

    overall_average_sit_to_stand = (
        average_sit_to_stand_per_user[
            "average_sit_to_stand_seconds"
        ]
        .mean()
    )


# =============================================================================
# 9. WALKING STEP DETECTION
# =============================================================================
#
# We analyse only samples labelled as:
#
#       walk
#
#
# Each continuous walking session is processed independently.
#
# This prevents a peak detector from incorrectly linking the end of one
# walking session with the beginning of another.
# =============================================================================


# -------------------------------------------------------------------------
# Import scipy's peak detection function.
# -------------------------------------------------------------------------

from scipy.signal import find_peaks


# -------------------------------------------------------------------------
# Extract walking data.
# -------------------------------------------------------------------------

walking_df = (
    df[
        df["activity"] == "walk"
    ]
    .copy()
)


print("\n")
print("=" * 80)
print("WALKING STEP DETECTION")
print("=" * 80)

print(
    f"\nNumber of walking samples: {len(walking_df):,}"
)

print(
    f"Number of walking sessions: "
    f"{walking_df['analysis_session_id'].nunique()}"
)


# =============================================================================
# 10. CONVERT MINIMUM STEP INTERVAL INTO SAMPLES
# =============================================================================

minimum_step_distance_samples = max(
    1,
    int(
        MIN_STEP_INTERVAL_SECONDS
        *
        SAMPLING_RATE_HZ
    )
)


# =============================================================================
# 11. DETECT STEPS WITHIN EACH WALKING SESSION
# =============================================================================
#
# Each walking session is analysed independently.
#
# The acceleration magnitude is first centred around its median:
#
#       dynamic_signal = acceleration - median(acceleration)
#
# A peak detection threshold is then calculated adaptively from the standard
# deviation of that session.
# =============================================================================

walking_session_records = []

walking_step_records = []


for (
    (current_user, current_session),
    walking_session
) in walking_df.groupby(
    [
        "user_id",
        "analysis_session_id"
    ]
):

    # ---------------------------------------------------------------------
    # Sort the session chronologically.
    # ---------------------------------------------------------------------

    walking_session = walking_session.sort_index()


    # ---------------------------------------------------------------------
    # Extract the acceleration magnitude signal.
    # ---------------------------------------------------------------------

    acceleration_signal = (
        walking_session["knee_accel_mag"]
        .to_numpy()
    )


    # ---------------------------------------------------------------------
    # Skip sessions that are too short.
    # ---------------------------------------------------------------------

    if len(acceleration_signal) < 3:

        continue


    # =====================================================================
    # REMOVE THE SESSION BASELINE
    # =====================================================================
    #
    # We subtract the median rather than integrating raw acceleration.
    #
    # This produces a signal centred around zero and makes the peak detector
    # focus more directly on repeated dynamic movement.
    # =====================================================================

    dynamic_acceleration = (
        acceleration_signal
        -
        np.median(
            acceleration_signal
        )
    )


    # =====================================================================
    # CALCULATE AN ADAPTIVE PEAK PROMINENCE
    # =====================================================================

    signal_std = np.std(
        dynamic_acceleration
    )


    peak_prominence = max(
        1e-6,
        STEP_PROMINENCE_MULTIPLIER
        *
        signal_std
    )


    # =====================================================================
    # DETECT PEAKS
    # =====================================================================
    #
    # distance ensures that two detected peaks cannot occur unrealistically
    # close together.
    #
    # prominence ensures that very small fluctuations are ignored.
    # =====================================================================

    peak_positions, peak_properties = find_peaks(
        dynamic_acceleration,
        distance=minimum_step_distance_samples,
        prominence=peak_prominence
    )


    # =====================================================================
    # CALCULATE STEP COUNT
    # =====================================================================
    #
    # Each detected gait-related peak is treated as one estimated step.
    #
    # This is an estimation assumption and should be validated against
    # ground-truth step counts if such data becomes available.
    # =====================================================================

    estimated_step_count = len(
        peak_positions
    )


    # =====================================================================
    # ESTIMATE WALKING DISTANCE
    # =====================================================================

    estimated_distance_meters = (
        estimated_step_count
        *
        ESTIMATED_STEP_LENGTH_METERS
    )


    # =====================================================================
    # CALCULATE WALKING SESSION DURATION
    # =====================================================================

    walking_duration_seconds = (
        len(walking_session)
        /
        SAMPLING_RATE_HZ
    )


    # ---------------------------------------------------------------------
    # Store walking session-level results.
    # ---------------------------------------------------------------------

    walking_session_records.append(
        {
            "user_id": current_user,

            "analysis_session_id": current_session,

            "walking_samples": len(
                walking_session
            ),

            "walking_duration_seconds": (
                walking_duration_seconds
            ),

            "estimated_step_count": (
                estimated_step_count
            ),

            "estimated_distance_meters": (
                estimated_distance_meters
            )
        }
    )


    # ---------------------------------------------------------------------
    # Store individual step information.
    # ---------------------------------------------------------------------

    session_indices = (
        walking_session.index.to_numpy()
    )


    for peak_position in peak_positions:

        walking_step_records.append(
            {
                "user_id": current_user,

                "analysis_session_id": current_session,

                "dataset_index": (
                    session_indices[
                        peak_position
                    ]
                )
            }
        )


# =============================================================================
# 12. CREATE WALKING RESULTS DATAFRAMES
# =============================================================================

walking_session_results = pd.DataFrame(
    walking_session_records
)


walking_step_results = pd.DataFrame(
    walking_step_records
)


# =============================================================================
# 13. CALCULATE WALKING DISTANCE PER INDIVIDUAL
# =============================================================================

if walking_session_results.empty:

    walking_distance_per_user = pd.DataFrame(
        columns=[
            "user_id",
            "number_of_walking_sessions",
            "total_estimated_steps",
            "total_estimated_distance_meters",
            "average_walking_session_distance_meters"
        ]
    )

    overall_average_walking_distance = np.nan


else:

    walking_distance_per_user = (
        walking_session_results
        .groupby("user_id")
        .agg(
            number_of_walking_sessions=(
                "analysis_session_id",
                "count"
            ),

            total_estimated_steps=(
                "estimated_step_count",
                "sum"
            ),

            total_estimated_distance_meters=(
                "estimated_distance_meters",
                "sum"
            ),

            average_walking_session_distance_meters=(
                "estimated_distance_meters",
                "mean"
            )
        )
        .reset_index()
    )


    # Calculate the average walking distance across individuals.

    overall_average_walking_distance = (
        walking_distance_per_user[
            "total_estimated_distance_meters"
        ]
        .mean()
    )


# =============================================================================
# 14. CREATE A COMBINED INDIVIDUAL SUMMARY
# =============================================================================

individual_summary = pd.merge(
    average_sit_to_stand_per_user,
    walking_distance_per_user,
    on="user_id",
    how="outer"
)


# =============================================================================
# 15. DISPLAY RESULTS
# =============================================================================

print("\n")
print("=" * 80)
print("SIT-TO-STAND RESULTS")
print("=" * 80)

display(
    average_sit_to_stand_per_user
)


if not np.isnan(overall_average_sit_to_stand):

    print(
        f"\nOverall average sit-to-stand time across individuals: "
        f"{overall_average_sit_to_stand:.3f} seconds"
    )


print("\n")
print("=" * 80)
print("ESTIMATED WALKING DISTANCE RESULTS")
print("=" * 80)

display(
    walking_distance_per_user
)


if not np.isnan(overall_average_walking_distance):

    print(
        f"\nAverage estimated walking distance across individuals: "
        f"{overall_average_walking_distance:.3f} metres"
    )


print("\n")
print("=" * 80)
print("COMBINED INDIVIDUAL SUMMARY")
print("=" * 80)

display(
    individual_summary
)


# =============================================================================
# END OF CELL 69
# =============================================================================
```

##### Average time taken to perform each activity
---

```python
# =============================================================================
# CELL 71: CALCULATE THE AVERAGE DURATION OF EACH ACTIVITY
# =============================================================================
#
# PURPOSE
# -------
#
# This cell calculates how long, on average, each activity is performed.
#
# The activities are analysed as individual continuous sessions rather than
# simply counting all samples belonging to an activity.
#
# For example:
#
#     Sitting  -> Sitting  -> Sitting
#
# followed by:
#
#     Walking  -> Walking
#
# followed by:
#
#     Sitting  -> Sitting
#
# represents THREE activity sessions:
#
#     Session 1: Sitting
#     Session 2: Walking
#     Session 3: Sitting
#
# The duration of each session is calculated using the sampling frequency:
#
#                     Number of Samples
#     Duration = -------------------------------
#                   Sampling Rate (20 Hz)
#
# Finally, the average activity duration is calculated across all detected
# sessions.
#
#
# IMPORTANT:
#
# The data prepared in Cell 68 is used here.
#
# Required variables:
#
#     df
#     SAMPLING_RATE_HZ
#
# =============================================================================


# =============================================================================
# 1. CHECK THAT THE REQUIRED VARIABLES EXIST
# =============================================================================

if "df" not in globals():

    raise NameError(
        "The DataFrame 'df' does not exist. "
        "Please run Cell 68 first."
    )


if "SAMPLING_RATE_HZ" not in globals():

    # -------------------------------------------------------------------------
    # Your desired sampling frequency.
    # -------------------------------------------------------------------------

    SAMPLING_RATE_HZ = 20


# =============================================================================
# 2. CREATE ACTIVITY SESSION IDENTIFIERS
# =============================================================================
#
# A new session begins when:
#
#   1. The user changes, OR
#   2. The activity changes.
#
# This is necessary because the same activity may occur multiple times
# for the same participant.
#
# Example:
#
#     Sitting
#     Sitting
#     Walking
#     Walking
#     Sitting
#
# The two Sitting periods should be treated as separate activity sessions.
# =============================================================================


df["activity_session_start"] = (
    (df["user_id"] != df["user_id"].shift())
    |
    (
        df["activity"]
        != df["activity"].shift()
    )
)


df["activity_session_id"] = (
    df["activity_session_start"]
    .cumsum()
)


# =============================================================================
# 3. CALCULATE THE DURATION OF EVERY INDIVIDUAL ACTIVITY SESSION
# =============================================================================
#
# For each:
#
#     user_id
#     activity_session_id
#
# we calculate:
#
#     activity
#     number of samples
#     duration in seconds
#
# =============================================================================


activity_session_durations = (
    df
    .groupby(
        [
            "user_id",
            "activity_session_id"
        ],
        as_index=False
    )
    .agg(
        activity=(
            "activity",
            "first"
        ),

        number_of_samples=(
            "activity",
            "size"
        )
    )
)


# =============================================================================
# 4. CONVERT NUMBER OF SAMPLES INTO TIME
# =============================================================================
#
# Since the sampling rate is:
#
#     20 samples per second
#
# then:
#
#     duration_seconds = number_of_samples / 20
# =============================================================================


activity_session_durations[
    "duration_seconds"
] = (
    activity_session_durations[
        "number_of_samples"
    ]
    /
    SAMPLING_RATE_HZ
)


# =============================================================================
# 5. CALCULATE AVERAGE ACTIVITY DURATION ACROSS ALL SESSIONS
# =============================================================================
#
# For each activity, calculate:
#
#     - Number of sessions
#     - Mean duration
#     - Median duration
#     - Standard deviation
#     - Minimum duration
#     - Maximum duration
#
# The median is included because it is less sensitive to unusually long
# or short activity sessions.
# =============================================================================


activity_duration_summary = (
    activity_session_durations
    .groupby(
        "activity",
        as_index=False
    )
    .agg(
        number_of_sessions=(
            "duration_seconds",
            "count"
        ),

        average_duration_seconds=(
            "duration_seconds",
            "mean"
        ),

        median_duration_seconds=(
            "duration_seconds",
            "median"
        ),

        standard_deviation_seconds=(
            "duration_seconds",
            "std"
        ),

        minimum_duration_seconds=(
            "duration_seconds",
            "min"
        ),

        maximum_duration_seconds=(
            "duration_seconds",
            "max"
        )
    )
)


# =============================================================================
# 6. CALCULATE AVERAGE DURATION PER USER AND ACTIVITY
# =============================================================================
#
# This prevents participants with many repeated sessions from dominating
# the overall result.
#
# First:
#
#     Calculate each participant's average duration for each activity.
#
# Then:
#
#     The overall average can be calculated from participant averages.
# =============================================================================


average_activity_duration_per_user = (
    activity_session_durations
    .groupby(
        [
            "user_id",
            "activity"
        ],
        as_index=False
    )
    .agg(
        number_of_sessions=(
            "duration_seconds",
            "count"
        ),

        average_duration_seconds=(
            "duration_seconds",
            "mean"
        )
    )
)


# =============================================================================
# 7. CALCULATE THE OVERALL AVERAGE ACROSS INDIVIDUALS
# =============================================================================
#
# Every participant is given equal weight.
#
# This is generally more appropriate when comparing human performance
# across participants.
# =============================================================================


average_activity_duration_across_individuals = (
    average_activity_duration_per_user
    .groupby(
        "activity",
        as_index=False
    )
    .agg(
        number_of_individuals=(
            "user_id",
            "nunique"
        ),

        average_duration_seconds=(
            "average_duration_seconds",
            "mean"
        ),

        standard_deviation_across_individuals=(
            "average_duration_seconds",
            "std"
        )
    )
)


# =============================================================================
# 8. SORT RESULTS
# =============================================================================

activity_duration_summary = (
    activity_duration_summary
    .sort_values(
        by="average_duration_seconds",
        ascending=False
    )
    .reset_index(drop=True)
)


average_activity_duration_across_individuals = (
    average_activity_duration_across_individuals
    .sort_values(
        by="average_duration_seconds",
        ascending=False
    )
    .reset_index(drop=True)
)


# =============================================================================
# 9. DISPLAY RESULTS
# =============================================================================

print("=" * 80)
print("AVERAGE DURATION OF EACH ACTIVITY")
print("=" * 80)


print(
    "\nAverage duration calculated across all activity sessions:"
)

display(
    activity_duration_summary
)


print(
    "\nAverage duration calculated across individuals:"
)

display(
    average_activity_duration_across_individuals
)


# =============================================================================
# END OF CELL 71
# =============================================================================
```

```python
# =============================================================================
# CELL 73: VISUAL VALIDATION OF WALKING STEP DETECTION
# =============================================================================
#
# PURPOSE
# -------
#
# The estimated walking distance calculated in Cell 69 depends directly on
# the accuracy of the step detection algorithm:
#
#     Estimated Distance = Detected Step Count × Estimated Step Length
#
# Therefore, before interpreting the estimated walking distance, it is
# important to visually inspect whether the detected peaks correspond to
# plausible walking steps.
#
#
# THIS CELL:
#
#   1. Allows a user and walking session to be selected.
#   2. Extracts the selected walking session.
#   3. Recreates the dynamic acceleration signal used for step detection.
#   4. Detects step peaks using the same parameters as Cell 69.
#   5. Plots:
#
#       - Raw knee acceleration magnitude.
#       - Dynamic knee acceleration.
#       - Detected step peaks.
#
#
# IMPORTANT
# ---------
#
# This cell is primarily a VALIDATION and VISUAL INSPECTION tool.
#
# The goal is to answer:
#
#     "Do the detected peaks appear to correspond to plausible walking steps?"
#
# =============================================================================


# =============================================================================
# 1. CHECK THAT THE REQUIRED DATA EXISTS
# =============================================================================

if "walking_session_results" not in globals():

    raise NameError(
        "walking_session_results does not exist. "
        "Please run Cell 69 first."
    )


if "walking_df" not in globals():

    raise NameError(
        "walking_df does not exist. "
        "Please run Cell 69 first."
    )


# =============================================================================
# 2. DISPLAY AVAILABLE WALKING SESSIONS
# =============================================================================
#
# This table allows you to see which user and session combinations are
# available for visual inspection.
# =============================================================================

print("=" * 80)
print("AVAILABLE WALKING SESSIONS")
print("=" * 80)

display(
    walking_session_results[
        [
            "user_id",
            "analysis_session_id",
            "walking_duration_seconds",
            "estimated_step_count",
            "estimated_distance_meters"
        ]
    ]
)


# =============================================================================
# 3. SELECT THE USER AND WALKING SESSION TO INSPECT
# =============================================================================
#
# Change these values after looking at the table displayed above.
#
# Example:
#
#     SELECTED_USER_ID = 1
#     SELECTED_WALKING_SESSION = 25
#
# =============================================================================

SELECTED_USER_ID = walking_session_results["user_id"].iloc[0]

SELECTED_WALKING_SESSION = (
    walking_session_results["analysis_session_id"].iloc[0]
)


# =============================================================================
# 4. EXTRACT THE SELECTED WALKING SESSION
# =============================================================================

selected_walking_session = (
    walking_df[
        (
            walking_df["user_id"] == SELECTED_USER_ID
        )
        &
        (
            walking_df["analysis_session_id"]
            == SELECTED_WALKING_SESSION
        )
    ]
    .copy()
)


# Ensure chronological ordering.

selected_walking_session = (
    selected_walking_session
    .sort_index()
)


if selected_walking_session.empty:

    raise ValueError(
        "The selected user/session combination does not exist."
    )


# =============================================================================
# 5. CREATE A TIME AXIS
# =============================================================================
#
# Since the IMU sampling frequency is 20 Hz:
#
#     time = sample number / 20
# =============================================================================

time_seconds = (
    np.arange(
        len(selected_walking_session)
    )
    /
    SAMPLING_RATE_HZ
)


# =============================================================================
# 6. EXTRACT RAW KNEE ACCELERATION MAGNITUDE
# =============================================================================

raw_acceleration = (
    selected_walking_session["knee_accel_mag"]
    .to_numpy()
)


# =============================================================================
# 7. RECREATE THE DYNAMIC ACCELERATION SIGNAL
# =============================================================================
#
# This uses the same baseline-centering approach as Cell 69.
#
# The median is removed so that the signal oscillates around zero.
# =============================================================================

dynamic_acceleration = (
    raw_acceleration
    -
    np.median(raw_acceleration)
)


# =============================================================================
# 8. RECREATE THE PEAK DETECTION PARAMETERS
# =============================================================================

signal_std = np.std(
    dynamic_acceleration
)


peak_prominence = max(
    1e-6,
    STEP_PROMINENCE_MULTIPLIER
    *
    signal_std
)


minimum_step_distance_samples = max(
    1,
    int(
        MIN_STEP_INTERVAL_SECONDS
        *
        SAMPLING_RATE_HZ
    )
)


# =============================================================================
# 9. DETECT STEP PEAKS
# =============================================================================

peak_positions, peak_properties = find_peaks(
    dynamic_acceleration,
    distance=minimum_step_distance_samples,
    prominence=peak_prominence
)


# =============================================================================
# 10. DISPLAY SESSION INFORMATION
# =============================================================================

estimated_steps = len(
    peak_positions
)


estimated_distance = (
    estimated_steps
    *
    ESTIMATED_STEP_LENGTH_METERS
)


print("\n")
print("=" * 80)
print("SELECTED WALKING SESSION")
print("=" * 80)

print(
    f"User ID: {SELECTED_USER_ID}"
)

print(
    f"Walking Session ID: {SELECTED_WALKING_SESSION}"
)

print(
    f"Session Duration: "
    f"{len(selected_walking_session) / SAMPLING_RATE_HZ:.2f} seconds"
)

print(
    f"Detected Steps: {estimated_steps}"
)

print(
    f"Estimated Distance: "
    f"{estimated_distance:.2f} metres"
)


# =============================================================================
# 11. PLOT RAW KNEE ACCELERATION MAGNITUDE
# =============================================================================

plt.figure(
    figsize=(15, 6)
)

plt.plot(
    time_seconds,
    raw_acceleration
)

plt.xlabel(
    "Time (seconds)"
)

plt.ylabel(
    "Knee Acceleration Magnitude"
)

plt.title(
    "Raw Knee Acceleration During the Selected Walking Session"
)

plt.grid(
    alpha=0.3
)

plt.tight_layout()

plt.show()


# =============================================================================
# 12. PLOT DYNAMIC ACCELERATION WITH DETECTED STEPS
# =============================================================================

plt.figure(
    figsize=(15, 6)
)

plt.plot(
    time_seconds,
    dynamic_acceleration,
    label="Dynamic Knee Acceleration"
)


plt.scatter(
    time_seconds[peak_positions],
    dynamic_acceleration[peak_positions],
    label="Detected Steps",
    zorder=3
)


plt.xlabel(
    "Time (seconds)"
)

plt.ylabel(
    "Dynamic Knee Acceleration"
)

plt.title(
    "IMU-Based Walking Step Detection from Knee Acceleration"
)

plt.legend()

plt.grid(
    alpha=0.3
)

plt.tight_layout()

plt.show()


# =============================================================================
# END OF CELL 73
# =============================================================================
```

```python
# =============================================================================
# CELL 74: VISUAL VALIDATION OF SIT-TO-STAND TRANSITION DETECTION
# =============================================================================
#
# PURPOSE
# -------
#
# Cell 69 estimated sit-to-stand duration by identifying periods of increased
# movement variance around labelled sit -> stand transitions.
#
# This cell allows an individual transition to be inspected visually.
#
#
# THE PLOT SHOWS:
#
#   1. Knee acceleration magnitude.
#   2. Rolling motion variance.
#   3. Adaptive movement threshold.
#   4. Detected movement start.
#   5. Detected movement end.
#   6. Labelled sit -> stand transition boundary.
#
#
# This provides an important quality-control check:
#
#     Is the detected movement period actually associated with the
#     sit-to-stand transition?
#
# =============================================================================


# =============================================================================
# 1. CHECK THAT RESULTS EXIST
# =============================================================================

if "sit_to_stand_results" not in globals():

    raise NameError(
        "sit_to_stand_results does not exist. "
        "Please run Cell 69 first."
    )


if sit_to_stand_results.empty:

    raise ValueError(
        "No valid sit-to-stand transitions were detected."
    )


# =============================================================================
# 2. DISPLAY AVAILABLE TRANSITIONS
# =============================================================================

print("=" * 80)
print("AVAILABLE SIT-TO-STAND TRANSITIONS")
print("=" * 80)

display(
    sit_to_stand_results[
        [
            "user_id",
            "transition_index",
            "movement_start_index",
            "movement_end_index",
            "sit_to_stand_seconds"
        ]
    ]
)


# =============================================================================
# 3. SELECT A TRANSITION TO INSPECT
# =============================================================================

SELECTED_TRANSITION_NUMBER = 0


if (
    SELECTED_TRANSITION_NUMBER
    >=
    len(sit_to_stand_results)
):

    raise ValueError(
        "SELECTED_TRANSITION_NUMBER is outside the available range."
    )


selected_transition = (
    sit_to_stand_results
    .iloc[
        SELECTED_TRANSITION_NUMBER
    ]
)


SELECTED_USER_ID = (
    selected_transition["user_id"]
)


TRANSITION_INDEX = int(
    selected_transition["transition_index"]
)


MOVEMENT_START_INDEX = int(
    selected_transition["movement_start_index"]
)


MOVEMENT_END_INDEX = int(
    selected_transition["movement_end_index"]
)


# =============================================================================
# 4. DEFINE A VISUALISATION WINDOW
# =============================================================================
#
# We display several seconds before and after the transition.
# =============================================================================

VISUAL_WINDOW_SECONDS = 6


half_window_samples = int(
    (
        VISUAL_WINDOW_SECONDS
        /
        2
    )
    *
    SAMPLING_RATE_HZ
)


start_index = max(
    0,
    TRANSITION_INDEX - half_window_samples
)


end_index = min(
    len(df) - 1,
    TRANSITION_INDEX + half_window_samples
)


visual_window = (
    df
    .loc[
        start_index:end_index
    ]
    .copy()
)


# Keep only the selected participant.

visual_window = (
    visual_window[
        visual_window["user_id"]
        ==
        SELECTED_USER_ID
    ]
    .copy()
)


# =============================================================================
# 5. CALCULATE MOTION VARIANCE
# =============================================================================

variance_window_samples = max(
    3,
    int(
        VARIANCE_WINDOW_SECONDS
        *
        SAMPLING_RATE_HZ
    )
)


visual_window["motion_variance"] = (
    visual_window["knee_accel_mag"]
    .rolling(
        window=variance_window_samples,
        center=True,
        min_periods=1
    )
    .var()
    .fillna(0)
)


# =============================================================================
# 6. CREATE A TIME AXIS
# =============================================================================

time_seconds = (
    (
        visual_window.index
        -
        TRANSITION_INDEX
    )
    /
    SAMPLING_RATE_HZ
)


# =============================================================================
# 7. PLOT KNEE ACCELERATION MAGNITUDE
# =============================================================================

plt.figure(
    figsize=(24, 6)
)

plt.plot(
    time_seconds,
    visual_window["knee_accel_mag"]
)


plt.axvline(
    0,
    linestyle="--",
    label="Labelled Sit-to-Stand Boundary"
)


plt.axvline(
    (
        MOVEMENT_START_INDEX
        -
        TRANSITION_INDEX
    )
    /
    SAMPLING_RATE_HZ,
    linestyle="--",
    label="Detected Movement Start"
)


plt.axvline(
    (
        MOVEMENT_END_INDEX
        -
        TRANSITION_INDEX
    )
    /
    SAMPLING_RATE_HZ,
    linestyle="--",
    label="Detected Movement End"
)


plt.xlabel(
    "Time Relative to Activity Boundary (seconds)"
)

plt.ylabel(
    "Knee Acceleration Magnitude"
)

plt.title(
    "Knee Motion During a Detected Sit-to-Stand Transition"
)

plt.legend()

plt.grid(
    alpha=0.3
)

plt.tight_layout()

plt.show()


# =============================================================================
# 8. PLOT MOTION VARIANCE AND ADAPTIVE THRESHOLD
# =============================================================================

plt.figure(
    figsize=(24, 6)
)

plt.plot(
    time_seconds,
    visual_window["motion_variance"],
    label="Rolling Motion Variance"
)


plt.axhline(
    selected_transition["movement_threshold"],
    linestyle="--",
    label="Adaptive Movement Threshold"
)


plt.axvline(
    0,
    linestyle="--",
    label="Labelled Sit-to-Stand Boundary"
)


plt.axvline(
    (
        MOVEMENT_START_INDEX
        -
        TRANSITION_INDEX
    )
    /
    SAMPLING_RATE_HZ,
    linestyle="--",
    label="Detected Movement Start"
)


plt.axvline(
    (
        MOVEMENT_END_INDEX
        -
        TRANSITION_INDEX
    )
    /
    SAMPLING_RATE_HZ,
    linestyle="--",
    label="Detected Movement End"
)


plt.xlabel(
    "Time Relative to Activity Boundary (seconds)"
)

plt.ylabel(
    "Motion Variance"
)

plt.title(
    "Detection of Active Movement During a Sit-to-Stand Transition"
)

plt.legend()

plt.grid(
    alpha=0.3
)

plt.tight_layout()

plt.show()


# =============================================================================
# 9. DISPLAY DETECTED TRANSITION INFORMATION
# =============================================================================

print("=" * 80)
print("SIT-TO-STAND TRANSITION SUMMARY")
print("=" * 80)

print(
    f"User ID: {SELECTED_USER_ID}"
)

print(
    f"Transition Index: {TRANSITION_INDEX}"
)

print(
    f"Detected Movement Duration: "
    f"{selected_transition['sit_to_stand_seconds']:.3f} seconds"
)


# =============================================================================
# END OF CELL 74
# =============================================================================
```

```python
# =============================================================================
# CELL 77: FEATURE CORRELATION AND REDUNDANCY ANALYSIS
# =============================================================================
#
# PURPOSE
# -------
#
# Many engineered features may contain similar information.
#
# For example:
#
#     knee_accel_energy
#
# may be strongly related to:
#
#     knee_accel_mag
#
# Highly correlated features can introduce redundancy into a machine-learning
# model.
#
# This cell:
#
#   1. Calculates the feature correlation matrix.
#   2. Visualises the correlation matrix.
#   3. Identifies highly correlated feature pairs.
#
# =============================================================================


# =============================================================================
# 1. SELECT NUMERIC FEATURES
# =============================================================================

correlation_features = pd.DataFrame(
    feature_engineered_dataset_a.select_dtypes(include='float64')
)


print(" ---------------------------------------------->>>> ------>>> ")
correlation_features.head()
print(" ---------------------------------------------->>>> ------>>> ")


# =============================================================================
# 2. CALCULATE THE CORRELATION MATRIX
# =============================================================================

correlation_matrix = (
    correlation_features
    .corr()
)


# =============================================================================
# 3. DISPLAY THE CORRELATION MATRIX
# =============================================================================

display(
    correlation_matrix
)


# =============================================================================
# 4. VISUALISE THE CORRELATION MATRIX
# =============================================================================

plt.figure(
    figsize=(24, 18)
)


plt.imshow(
    correlation_matrix,
    aspect="auto"
)


plt.colorbar(
    label="Correlation Coefficient"
)


plt.xticks(
    range(
        len(correlation_matrix.columns)
    ),
    correlation_matrix.columns,
    rotation=90
)


plt.yticks(
    range(
        len(correlation_matrix.index)
    ),
    correlation_matrix.index
)


plt.title(
    "Correlation Matrix of Engineered IMU Features"
)


plt.tight_layout()

plt.show()


# =============================================================================
# 5. IDENTIFY HIGHLY CORRELATED FEATURE PAIRS
# =============================================================================
#
# A correlation threshold of 0.90 is used as an initial screening criterion.
#
# This does not automatically mean one feature must be removed.
#
# Instead, it identifies features that may contain highly overlapping
# information and should be considered during feature selection.
# =============================================================================

CORRELATION_THRESHOLD = 0.90


highly_correlated_pairs = []


for i in range(
    len(correlation_matrix.columns)
):

    for j in range(
        i + 1,
        len(correlation_matrix.columns)
    ):

        correlation_value = (
            correlation_matrix
            .iloc[i, j]
        )


        if (
            abs(correlation_value)
            >=
            CORRELATION_THRESHOLD
        ):

            highly_correlated_pairs.append(
                {
                    "feature_1": (
                        correlation_matrix.columns[i]
                    ),

                    "feature_2": (
                        correlation_matrix.columns[j]
                    ),

                    "correlation": (
                        correlation_value
                    ),

                    "absolute_correlation": (
                        abs(correlation_value)
                    )
                }
            )


highly_correlated_features = (
    pd.DataFrame(
        highly_correlated_pairs
    )
)


if not highly_correlated_features.empty:

    highly_correlated_features = (
        highly_correlated_features
        .sort_values(
            by="absolute_correlation",
            ascending=False
        )
        .reset_index(drop=True)
    )


print("=" * 80)
print(
    f"HIGHLY CORRELATED FEATURE PAIRS "
    f"(|r| >= {CORRELATION_THRESHOLD})"
)
print("=" * 80)


if highly_correlated_features.empty:

    print(
        "No highly correlated feature pairs were identified."
    )

else:

    display(
        highly_correlated_features
    )


# =============================================================================
# END OF CELL 77
# =============================================================================
```

##### Time Series Segmentation

```python
# =============================================================================
# CELL 82: REUSABLE TIME-SERIES SEGMENTATION FUNCTION
# =============================================================================
#
# PURPOSE
# -------
#
# Human Activity Recognition is fundamentally a time-series classification
# problem.
#
# A single IMU sample does not contain enough temporal information to fully
# describe an activity such as walking or jogging.
#
# Therefore, consecutive samples are grouped into fixed-length windows.
#
#
# EXAMPLE
# -------
#
# At 20 Hz, using a 2-second window:
#
#     2 seconds × 20 samples/second
#
#     = 40 samples per window
#
#
# The output shape will be:
#
#     (number_of_windows,
#      samples_per_window,
#      number_of_features)
#
#
# This function is reusable because the user can specify:
#
#     - DataFrame.
#     - Features.
#     - Activity classes.
#     - Window size.
#     - Window step.
#     - User column.
#     - Activity column.
#     - Session column.
#
# =============================================================================


def create_time_series_windows(
    dataframe,
    feature_columns,
    activity_classes,
    window_size_samples,
    window_step_samples,
    user_column="user_id",
    activity_column="activity",
    session_column=None
):
    """
    Converts continuous IMU data into fixed-length time-series windows.

    Parameters
    ----------
    dataframe : pandas.DataFrame
        The complete HAR dataset.

    feature_columns : list
        Names of the numerical sensor features that should be included
        in each time-series window.

    activity_classes : list
        Activity labels that should be included in the experiment.

    window_size_samples : int
        Number of consecutive IMU samples in each window.

    window_step_samples : int
        Number of samples to move forward before creating the next window.

        For example:

            window_size_samples = 40
            window_step_samples = 20

        creates a 50% overlap.

    user_column : str
        Column containing participant identifiers.

    activity_column : str
        Column containing normalised activity labels.

    session_column : str or None
        Optional session identifier.

        If provided, windows are created independently within each session.

        This prevents a window from accidentally combining samples from
        separate activity sessions.

    Returns
    -------
    X_windows : numpy.ndarray

        Shape:

            (
                number_of_windows,
                window_size_samples,
                number_of_features
            )

    y_windows : numpy.ndarray

        Activity label assigned to each window.

    window_users : numpy.ndarray

        The user associated with each window.

        This is required later for Leave-One-User-Out evaluation.

    window_sessions : numpy.ndarray

        Session identifier associated with each window.

        If no session column is provided, this contains NaN values.
    """


    # =========================================================================
    # 1. VALIDATE REQUIRED COLUMNS
    # =========================================================================

    required_columns = (
        list(feature_columns)
        +
        [
            user_column,
            activity_column
        ]
    )


    if session_column is not None:

        required_columns.append(
            session_column
        )


    missing_columns = [

        column

        for column in required_columns

        if column not in dataframe.columns

    ]


    if missing_columns:

        raise ValueError(
            "The following required columns are missing: "
            f"{missing_columns}"
        )


    # =========================================================================
    # 2. CREATE A WORKING COPY
    # =========================================================================

    working_data = dataframe.copy()


    # =========================================================================
    # 3. NORMALISE THE REQUESTED ACTIVITY CLASSES
    # =========================================================================

    requested_activities = [

        str(activity)
        .strip()
        .lower()

        for activity in activity_classes

    ]


    # =========================================================================
    # 4. KEEP ONLY THE REQUESTED ACTIVITIES
    # =========================================================================

    working_data = (
        working_data[
            working_data[activity_column]
            .isin(requested_activities)
        ]
        .copy()
    )


    # =========================================================================
    # 5. REMOVE ROWS WITH MISSING FEATURE VALUES
    # =========================================================================
    #
    # Windows containing missing sensor features cannot be used directly by
    # the models.
    # =========================================================================

    working_data = (
        working_data
        .replace(
            [
                np.inf,
                -np.inf
            ],
            np.nan
        )
    )


    working_data = (
        working_data
        .dropna(
            subset=feature_columns
        )
    )


    # =========================================================================
    # 6. DEFINE GROUPING STRUCTURE
    # =========================================================================
    #
    # Windows must never cross user boundaries.
    #
    # If session information is available, windows must also remain within
    # individual sessions.
    # =========================================================================

    grouping_columns = [

        user_column

    ]


    if session_column is not None:

        grouping_columns.append(
            session_column
        )


    # =========================================================================
    # 7. CREATE STORAGE FOR THE SEGMENTED DATA
    # =========================================================================

    all_windows = []

    all_window_labels = []

    all_window_users = []

    all_window_sessions = []


    # =========================================================================
    # 8. PROCESS EACH USER / SESSION INDEPENDENTLY
    # =========================================================================

    for group_identifiers, group_data in (
        working_data
        .groupby(
            grouping_columns,
            sort=False
        )
    ):


        # ---------------------------------------------------------------------
        # Extract the user ID.
        # ---------------------------------------------------------------------

        if session_column is None:

            current_user = group_identifiers

            current_session = np.nan

        else:

            current_user = group_identifiers[0]

            current_session = group_identifiers[1]


        # ---------------------------------------------------------------------
        # Activity changes are important.
        #
        # Even within the same user/session, we do not want a window that
        # combines:
        #
        #     Sitting + Standing
        #
        # or:
        #
        #     Walking + Jogging
        #
        # Therefore, the data is split into consecutive activity segments.
        # ---------------------------------------------------------------------

        activity_change = (
            group_data[activity_column]
            !=
            group_data[activity_column]
            .shift()
        )


        activity_segment_id = (
            activity_change
            .cumsum()
        )


        # ---------------------------------------------------------------------
        # Process each continuous activity segment independently.
        # ---------------------------------------------------------------------

        for _, activity_segment in (
            group_data
            .groupby(
                activity_segment_id,
                sort=False
            )
        ):


            # -------------------------------------------------------------
            # Obtain the activity label for this continuous segment.
            # -------------------------------------------------------------

            current_activity = (
                activity_segment[activity_column]
                .iloc[0]
            )


            # -------------------------------------------------------------
            # Extract numerical feature values.
            # -------------------------------------------------------------

            feature_values = (
                activity_segment[
                    feature_columns
                ]
                .to_numpy(
                    dtype=np.float32
                )
            )


            # -------------------------------------------------------------
            # Skip segments that are shorter than one complete window.
            # -------------------------------------------------------------

            if len(feature_values) < window_size_samples:

                continue


            # -------------------------------------------------------------
            # Slide the window across the activity segment.
            # -------------------------------------------------------------

            for start_position in range(
                0,
                len(feature_values)
                -
                window_size_samples
                +
                1,
                window_step_samples
            ):


                end_position = (
                    start_position
                    +
                    window_size_samples
                )


                # ---------------------------------------------------------
                # Extract one complete time-series window.
                # ---------------------------------------------------------

                current_window = (
                    feature_values[
                        start_position:end_position
                    ]
                )


                # ---------------------------------------------------------
                # Store the window and associated metadata.
                # ---------------------------------------------------------

                all_windows.append(
                    current_window
                )


                all_window_labels.append(
                    current_activity
                )


                all_window_users.append(
                    current_user
                )


                all_window_sessions.append(
                    current_session
                )


    # =========================================================================
    # 9. CONVERT LISTS TO NUMPY ARRAYS
    # =========================================================================

    X_windows = np.asarray(
        all_windows,
        dtype=np.float32
    )


    y_windows = np.asarray(
        all_window_labels
    )


    window_users = np.asarray(
        all_window_users
    )


    window_sessions = np.asarray(
        all_window_sessions
    )


    # =========================================================================
    # 10. VALIDATE THAT WINDOWS WERE CREATED
    # =========================================================================

    if len(X_windows) == 0:

        raise ValueError(
            "No time-series windows were created. "
            "Check the selected activities, session structure, "
            "window size, and available data."
        )


    return (
        X_windows,
        y_windows,
        window_users,
        window_sessions
    )
```

```python
# =============================================================================
# CELL 83: CREATE THE SEGMENTED HUMAN ACTIVITY RECOGNITION DATASET
# =============================================================================
#
# This cell applies the reusable segmentation function from Cell 82.
#
# The resulting dataset will contain:
#
#     X_time_series
#
#         IMU windows:
#
#         (
#             number_of_windows,
#             samples_per_window,
#             number_of_features
#         )
#
#
#     y_time_series
#
#         Activity label for each window.
#
#
#     time_series_users
#
#         User ID associated with each window.
#
#
#     time_series_sessions
#
#         Session ID associated with each window.
#
# =============================================================================


# =============================================================================
# 1. DETERMINE WHETHER A SESSION COLUMN EXISTS
# =============================================================================

if "session_id" in df.columns:

    SEGMENTATION_SESSION_COLUMN = "session_id"

else:

    SEGMENTATION_SESSION_COLUMN = None


# =============================================================================
# 2. CREATE TIME-SERIES WINDOWS
# =============================================================================

(
    X_time_series,
    y_time_series,
    time_series_users,
    time_series_sessions

) = create_time_series_windows(

    dataframe=feature_engineered_dataset_a,

    feature_columns=MODEL_FEATURES,

    activity_classes=ACTIVITIES_TO_MODEL,

    window_size_samples=WINDOW_SIZE_SAMPLES,

    window_step_samples=WINDOW_STEP_SAMPLES,

    user_column="user_id",

    activity_column="activity",

    session_column=SEGMENTATION_SESSION_COLUMN
)


# =============================================================================
# 3. DISPLAY DATASET SHAPE
# =============================================================================

print("=" * 80)
print("SEGMENTED HUMAN ACTIVITY RECOGNITION DATASET")
print("=" * 80)

print(
    f"\nNumber of Time-Series Windows: "
    f"{len(X_time_series):,}"
)

print(
    f"Samples Per Window: "
    f"{X_time_series.shape[1]}"
)

print(
    f"Features Per Sample: "
    f"{X_time_series.shape[2]}"
)

print(
    f"Number of Users: "
    f"{len(np.unique(time_series_users))}"
)


# =============================================================================
# 4. DISPLAY CLASS DISTRIBUTION
# =============================================================================

window_class_distribution = (
    pd.Series(
        y_time_series
    )
    .value_counts()
    .reset_index()
)


window_class_distribution.columns = [

    "activity",
    "number_of_windows"

]


window_class_distribution[
    "percentage"
] = (

    window_class_distribution[
        "number_of_windows"
    ]

    /

    len(y_time_series)

    *

    100

)


print("\nWINDOW-LEVEL CLASS DISTRIBUTION")

display(
    window_class_distribution
)


# =============================================================================
# 5. DISPLAY USER DISTRIBUTION
# =============================================================================

window_user_distribution = (
    pd.Series(
        time_series_users
    )
    .value_counts()
    .sort_index()
    .reset_index()
)


window_user_distribution.columns = [

    "user_id",
    "number_of_windows"

]


print("\nWINDOWS PER USER")

display(
    window_user_distribution
)
```

#### Timed-Up-and-Go (TUG)
---
The timed Up and Go (TUG) is a clinical test that evaluates functional mobility and fall risk. It measures the time it takes for a person to stand from a chair, walk threee meters(about 10 feet), turn around, walk back, and sit down again. 

The patient sits comfortably against a chair and rises on the word "Go", walks to a designated three-meter mark at a safe pace, turns around, and returns to seated. 

Equipment used include a standard arm chari, a stopwatch, a floor marker exactly three meters away and any assistive devices (eg walking aids)

Interpreting Results

The total time correlates with mobility and balance:Under 10 seconds: Normal mobility.11–20 seconds: Mostly independent with good functional mobility.Over 12–15 seconds: Often indicates an increased risk of falls and reduced functional independence.Over 30 seconds: Indicates significant mobility issues and a high risk of falling.
